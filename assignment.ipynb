```python
#                                      WEEK 1 

import statistics as st 

ages = [19,22,21,35,28,24,31,19,27,40];
x = len(ages);        #returns length of the list
y = sum(ages);        #returns sum of values in the list
z = round((y/x), 1);     #computes mean and rounds to 1 decimal place 
a = min(ages);        #returns the lowest value in the list
b = max(ages);        #returns the highest value in the list
c = st.median(ages);     #returns the median value in the list


print(f"There are {x} people in the survey");
print(f"The total of all ages is {y}");
print(f"Mean age to one decimal place is {z}");
print(f"Youngest age is {a}");
print(f"Oldest age is {b}");
print(f"Median age is {c}");

```

```python
#                                  WEEK 3

import pandas as pd
df = pd.read_excel("50_startups_messy.xlsx")

print("================================================================================================================================================================")
print(df.shape)                   #prints the shape of the dataset
print(df.isna().sum())            #prints the counts the number of missing (NaN) values in each column of the DataFrame
print(df.duplicated().sum())      #prints the number of duplicate rows in the entire dataset
print("================================================================================================================================================================")
df = df.drop_duplicates()                                   #gets rid of all the duplicates
df["State"] = df["State" ].str.strip().str.title()          #removes trailing spaces in the text and formats them to lowercase for uniformity
df["Marketing Spend" ] = pd.to_numeric(df["Marketing Spend"], errors="coerce" )     #changes the 'Marketing Spend' column type to numeric values

print("")
print(df["State"])
print("")
print(df["Marketing Spend"])
print("================================================================================================================================================================")
srs = df.sample(n = 20, random_state = 7)           #draws a simple random sample from the dataset
strat = df.groupby("State", group_keys=False ).apply(lambda g: g.sample(frac=0.2))     #draws a stratified sample with the 'State' column as the strata
print("")
print(srs)
print("")
print(strat)
print("================================================================================================================================================================")
compare_RD_Spend = pd.DataFrame({"population":df["R&D Spend"].value_counts(normalize=True ),"sample": srs["R&D Spend"].value_counts(normalize=True )}) 
compare_Marketing_Spend = pd.DataFrame({"population":df["Marketing Spend"].value_counts(normalize=True ),"sample": srs["Marketing Spend"].value_counts(normalize=True )}) 

print("")
print(compare_RD_Spend.round(4))             #compares the mean score of R&D Spend in both samples against the population.
print("")
print(compare_Marketing_Spend.round(4))      #Compare the mean score Marketing Spend in both samples against the population.
print("================================================================================================================================================================")
print("The sample for R&D Spend is closer to the population")
print("================================================================================================================================================================")
```
